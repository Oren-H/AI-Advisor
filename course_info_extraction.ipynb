```python
from IPython.display import Markdown, display
from openai import OpenAI
from pprint import pprint

client = OpenAI()
```

## Schema

```python
from pydantic import BaseModel, Field
from typing import Optional, List, Literal
```

```python
DayCode = Literal["M", "T", "W", "TR", "F", "S", "SU"]

class Course(BaseModel):
    title: str
    course_number: str
    description: Optional[str] = None
    section_call_number: Optional[str] = None
    days: Optional[List[DayCode]] = None
    start_time: Optional[str] = None
    end_time: Optional[str] = None
    location: Optional[str] = None
    instructor: Optional[str] = None
    points_credits: Optional[str] = None
    enrollment: Optional[str] = None
    prerequisites: Optional[str] = None
    link: Optional[str] = None

class AdvisoryResponse(BaseModel):
    reply: str = Field(description="Natural language answer to the user")
    courses: List[Course] = Field(description="Zero or more courses referenced in the reply, in structured format")
```

```python
user_prompt = "What intro computer science courses should I take at Columbia University?"
system_prompt = """You are a Columbia University college advisor. 
Answer the user's questions by searching the web to find valid course offerings from the fall 2025 term.
Only use trusted, official college webistes to find course offerings. 

First, you will compile a list of relevant course offerings. For each course you find that you think is relevant to the user's question, you will note the following:
1. course title
2. course number
3. description
4. Section/Call number
5. days
6. times
6. Location
7. instructor
8. points/credits
9. enrollment
10. Prerequisites

In a structured output, you will return both a natural language response to the user's question and a structued list of course information for each course you reference. 
IMPORTANT:If a course has multiple sections, make sure the course list contains a distinct course object for each section.
IMPORTANT: Don't include inline citations in your response. 

"""
```

```python
response = client.responses.parse(
    model="gpt-4o-mini",
    tools=[{"type": "web_search_preview",
            "search_context_size": "low"}],
    input=[
        {"role": "system", "content": system_prompt},   
        {"role": "user", "content": user_prompt}
    ],
    text_format=AdvisoryResponse
)
```

```python
display(Markdown(response.output_parsed.reply))
```

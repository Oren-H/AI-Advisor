````python
# Conversation state
history_unstructured = [
    {"role": "system", "content": "You are a helpful academic advisor for Columbia University. Cite sources when possible."}
]
model = "gpt-5"
tools = [{"type": "web_search_preview"}]

def chat_unstructured(user_input: str):
    global history_unstructured
    convo = history_unstructured + [{"role": "user", "content": user_input}]
    resp = client.responses.create(
        model=model,
        tools=tools,
        input=convo  # Use conversation as input for multi-turn
    )
    text = resp.output_text
    history_unstructured += [
        {"role": "user", "content": user_input},
        {"role": "assistant", "content": text},
    ]
    display(Markdown(text))
    return text
````

````python
from pydantic import BaseModel, Field
from typing import Optional, List, Literal
````

````python
DayCode = Literal["M", "T", "W", "TR", "F", "S", "SU"]

class Course(BaseModel):
    title: str
    course_number: str
    description: Optional[str] = None
    section_call_number: Optional[str] = None
    days: Optional[List[DayCode]] = None
    start_time: Optional[str] = None
    end_time: Optional[str] = None
    location: Optional[str] = None
    instructor: Optional[str] = None
    points_credits: Optional[str] = None
    enrollment: Optional[str] = None
    prerequisites: Optional[str] = None
    link: Optional[str] = None

class AdvisoryResponse(BaseModel):
    school_wide_requirements: str = Field(description="Summary of school-wide requirements. This changes based on the user's school (CC, GS, SEAS, etc.)")
    major_requirements: str = Field(description="Summary of major requirements")
    schedule: str = Field(description="Custom schedule for the user that satisfies the major requirements and the user's preferences")
    reasoning: str = Field(description="A natural language response to the user's question that includes the reasoning behind the schedule")
````

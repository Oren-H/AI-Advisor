```python
# Conversation state
history_unstructured = [
    {"role": "system", "content": "You are a helpful academic advisor for Columbia University. Cite sources when possible."}
]
model = "gpt-5"
tools = [{"type": "web_search_preview"}]

def chat_unstructured(user_input: str):
    global history_unstructured
    convo = history_unstructured + [{"role": "user", "content": user_input}]
    resp = client.responses.create(
        model=model,
        tools=tools,
        input=convo  # Use conversation as input for multi-turn
    )
    text = resp.output_text
    history_unstructured += [
        {"role": "user", "content": user_input},
        {"role": "assistant", "content": text},
    ]
    display(Markdown(text))
    return text
```

```python
from pydantic import BaseModel, Field
from typing import Optional, List, Literal
```

```python
DayCode = Literal["M", "T", "W", "TR", "F", "S", "SU"]

class Course(BaseModel):
    title: str
    course_number: str
    description: Optional[str] = None
    section_call_number: Optional[str] = None
    days: Optional[List[DayCode]] = None
    start_time: Optional[str] = None
    end_time: Optional[str] = None
    location: Optional[str] = None
    instructor: Optional[str] = None
    points_credits: Optional[str] = None
    enrollment: Optional[str] = None
    prerequisites: Optional[str] = None
    link: Optional[str] = None

class AdvisoryResponse(BaseModel):
    reply: str = Field(description="Natural language answer to the user")
    courses: List[Course] = Field(description="Zero or more courses referenced in the reply, in structured format")
```

```python
# Conversation state

system_prompt = """You are a Columbia University college advisor. 
Answer the user's questions by searching the web to find valid major requirements andcourse offerings from the fall 2025 term.
Only use trusted, official college webistes to find major and course offerings. 

IMPORTANT: Don't include inline citations in your response. 
"""

history = [
    {"role": "system", "content": system_prompt}

]
model = "gpt-5"
tools = [{"type": "web_search_preview",
            "search_context_size": "low"}]

def chat_structured(user_input: str):
    global history
    convo = history + [{"role": "user", "content": user_input}]
    resp = client.responses.parse(
        model=model,
        tools=tools,
        input=convo,
        text_format = AdvisoryResponse
    )
    
    history += [
        {"role": "user", "content": user_input},
        {"role": "assistant", "content": resp.output_parsed.reply},
    ]
    display(Markdown(resp.output_parsed.reply))
    return resp.output_parsed
```

```python
output = chat_structured("""
Could you give me a 4 year plan for a Columbia University CC first-year student? I want
to major in Applied Math and minor in Computer Science 
Here is what I have taken in highschool:
AP Computer Science A
AP Calculus BC
AP Physics 1
AP Physics 2
AP Chemistry
AP Biology
AP English Language and Composition
AP English Literature and Composition
AP History
AP Government and Politics
AP Economics
AP Spanish Language and Culture

I want to go into quant trading. I am interested in simulation, statistics, optimization,
and machine learning. 

""")
```
